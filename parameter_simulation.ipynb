```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yaml
from tqdm import tqdm
import warnings
warnings.filterwarnings("ignore")

from server import run_simulation
```

```python
def run_param_sweep(config, results_path):
    with open(results_path, "w") as f:
        f.write(
            "num_ugvs,ugv_range,maze_loop_percentage,"
            "tasks_completed,distance_traveled,energy_consumed,avg_task_completion_time\n"
        )
        # nested tqdm bars to monitor progress
        for num_ugvs in tqdm(range(3, 8), desc="Num UGVs"):
            for maze_loop_percentage in tqdm(range(50, 91, 10),
                                             desc="Maze Loop %",
                                             leave=False):
                for ugv_range in tqdm(range(1500, 5001, 500),
                                      desc="UGV Range",
                                      leave=False):
                    print("Num UGVs, UGV Range, Maze LP:", num_ugvs, ugv_range, maze_loop_percentage)
                    config["world"]["num_ugvs"] = num_ugvs
                    config["ugv"]["range"] = ugv_range
                    config["world"]["maze_loop_percentage"] = maze_loop_percentage

                    results = run_simulation(config,
                                             render_mode="quiet",
                                             env_type="multi")
                    tasks_completed = results["tasks_completed"]
                    distance_traveled = results["distance_traveled"]
                    energy_consumed = results["energy_consumed"]
                    avg_task_completion_time = results["avg_task_completion_time"]

                    results_line = (
                        f"{num_ugvs}, {ugv_range}, {maze_loop_percentage}, "
                        f"{tasks_completed}, {distance_traveled}, "
                        f"{energy_consumed}, {avg_task_completion_time}\n"
                    )
                    f.write(results_line)
```

```python
config_file = f"runs/dijkstras/config.yaml"

config = yaml.safe_load(open(config_file, "r"))
results_path = f"runs/dijkstras/param_sweep_no_bms.csv"
run_param_sweep(config, results_path)
```

```python
config['world']['bms'] = True
results_path = f"runs/dijkstras/param_sweep_bms.csv"
run_param_sweep(config, results_path)
```

```python
policy_name = "9e19ef9b"
config_file = f"runs/sarsa/{policy_name}/config.yaml"

config = yaml.safe_load(open(config_file, "r"))
results_path = f"runs/sarsa/{policy_name}/param_sweep_no_bms.csv"
run_param_sweep(config, results_path)
```

```python
config['world']['bms'] = True
results_path = f"runs/sarsa/{policy_name}/param_sweep_bms.csv"
run_param_sweep(config, results_path)
```

```python
def plot_param(results_d_nb, results_d_b, results_s_nb, results_s_b, param):
    grouped_d_nb = results_d_nb.groupby(param).mean().reset_index()
    grouped_d_b = results_d_b.groupby(param).mean().reset_index()
    grouped_s_nb = results_s_nb.groupby(param).mean().reset_index()
    grouped_s_b = results_s_b.groupby(param).mean().reset_index()

    fig, axes = plt.subplots(3, 1, figsize=(8, 12), sharex=True)

    axes[0].plot(grouped_d_nb[param], grouped_d_nb['tasks_completed'], label='Dijkstra No BMS', color='red', linestyle='--')
    axes[0].plot(grouped_d_b[param], grouped_d_b['tasks_completed'], label='Dijkstra BMS', color='red')
    axes[0].plot(grouped_s_nb[param], grouped_s_nb['tasks_completed'], label='SARSA No BMS', color='blue', linestyle='--')
    axes[0].plot(grouped_s_b[param], grouped_s_b['tasks_completed'], label='SARSA BMS', color='blue')
    axes[0].set_title(f'Tasks Completed vs {param}')
    axes[0].set_ylabel('Tasks Completed')
    axes[0].legend()

    axes[1].plot(grouped_d_nb[param], grouped_d_nb['energy_consumed'], label='Dijkstra No BMS', color='red', linestyle='--')
    axes[1].plot(grouped_d_b[param], grouped_d_b['energy_consumed'], label='Dijkstra BMS', color='red')
    axes[1].plot(grouped_s_nb[param], grouped_s_nb['energy_consumed'], label='SARSA No BMS', color='blue', linestyle='--')
    axes[1].plot(grouped_s_b[param], grouped_s_b['energy_consumed'], label='SARSA BMS', color='blue')
    axes[1].set_title(f'Energy Consumed vs {param}')
    axes[1].set_ylabel('Energy Consumed')
    axes[1].legend()

    axes[2].plot(grouped_d_nb[param], grouped_d_nb['avg_task_completion_time'], label='Dijkstra No BMS', color='red', linestyle='--')
    axes[2].plot(grouped_d_b[param], grouped_d_b['avg_task_completion_time'], label='Dijkstra BMS', color='red')
    axes[2].plot(grouped_s_nb[param], grouped_s_nb['avg_task_completion_time'], label='SARSA No BMS', color='blue', linestyle='--')
    axes[2].plot(grouped_s_b[param], grouped_s_b['avg_task_completion_time'], label='SARSA BMS', color='blue')
    axes[2].set_title(f'Task Completion Time vs {param}')
    axes[2].set_ylabel('Task Completion Time')
    axes[2].legend()

    plt.tight_layout()
    plt.show()
```

```python
dij_nb_df = pd.read_csv("runs/dijkstras/param_sweep_no_bms.csv")
dij_b_df = pd.read_csv("runs/dijkstras/param_sweep_bms.csv")
sarsa_nb_df = pd.read_csv(f"runs/sarsa/{policy_name}/param_sweep_no_bms.csv")
sarsa_b_df = pd.read_csv(f"runs/sarsa/{policy_name}/param_sweep_bms.csv")
plot_param(dij_nb_df, dij_b_df, sarsa_nb_df, sarsa_b_df, "num_ugvs")
```
